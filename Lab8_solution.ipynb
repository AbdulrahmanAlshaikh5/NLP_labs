### pos_tag method

````python
import nltk
from nltk import pos_tag
from nltk.tokenize import word_tokenize, sent_tokenize

resources = {
    "punkt_tab": "tokenizers/punkt_tab/english/",
    "averaged_perceptron_tagger_eng": "taggers/averaged_perceptron_tagger_eng/"
}
for package, resource_path in resources.items():
    try:
        nltk.data.find(resource_path)
    except LookupError:
        if not nltk.download(package, quiet=True):
            raise RuntimeError(f"Could not download NLTK resource: {package}")
print("NLTK resources are ready.")
````

````python
text = "I received my new phone broken from Amazon."
tokens = word_tokenize(text)
tagged_tokens = pos_tag(tokens)
for word, tag in tagged_tokens:
    print(f"Word: {word:<12} POS tag: {tag}")
````

````python
import spacy

if not spacy.util.is_package("en_core_web_sm"):
    subprocess.run([sys.executable, "-m", "spacy", "download", "en_core_web_sm"], check=True)
nlp = spacy.load("en_core_web_sm")
print("spaCy version:", spacy.__version__)
print("English model version:", nlp.meta["version"])
````

````python
doc = nlp("The quick brown fox jumped over the lazy dog's back.")
print("Text:", doc.text)
````

* To view the coarse POS tag use `token.pos_`
* To view the fine-grained tag use `token.tag_`
* To view the description of either type of tag use `spacy.explain(tag)`

````python
for token in doc:
    print(f'{token.text:{10}} {token.pos_:{8}} {token.tag_:{6}} {spacy.explain(token.tag_)}')
````

In the English language, the same word can have different meanings, even within the same sentence.

For this reason, morphology is important. **spaCy** uses machine learning algorithms to best predict the use of a token in a sentence. Is *"I read books on NLP"* present or past tense?

The spelling of “read” is the same in present and past tense. The following examples show the model’s predictions; a sentence without a time cue can remain grammatically ambiguous.

````python
doc = nlp(u'I read books on NLP.')
r = doc[1]

print(f'{r.text:{10}} {r.pos_:{8}} {r.tag_:{6}} {spacy.explain(r.tag_)}')
````

````python
doc = nlp(u'I read a book on NLP.')
r = doc[1]

print(f'{r.text:{10}} {r.pos_:{8}} {r.tag_:{6}} {spacy.explain(r.tag_)}')
````


What tag will be assigned if you added **'always'**?

“Always” suggests a habitual action, so “read” is interpreted as present tense: `VBP` (verb, non-3rd person singular present). The next cell shows the actual model prediction.

````python
doc = nlp(u'I always read a book on NLP.')
r = doc[2]

print(f'{r.text:{10}} {r.pos_:{8}} {r.tag_:{6}} {spacy.explain(r.tag_)}')
````

## Example: Find the adjective words in a sentance


````python
from nltk import pos_tag
from nltk.tokenize import word_tokenize, sent_tokenize

print("NLTK tokenization and tagging functions are ready.")
````

````python
text = "Mobile is very handy. Packed with good performances and nice camera. This is my first iPhone. Very good experience with Apple. Battery is worst. I am hardly getting 5 hours screen on time. Positive thing is I came to know how many hours I was spending on my mobile. Because of poor battery now I am only using mobile for 4 hours a day."
for sentence in sent_tokenize(text):
    tagged_sentence = pos_tag(word_tokenize(sentence))
    adjectives = [(word, tag) for word, tag in tagged_sentence if tag.startswith("JJ")]
    print("Sentence:", sentence)
    print("Adjectives:", adjectives)
    print()
````

### Task 2

Compare the explanation of the words “google”  and “table” in the following sentences.
1. “Did you search on Google?”
2. “Let us google it.”
3. “The has placed the food on the table”
4. “She plans to table a motion for debate in Parliament”

**Hint:**
`word = sen[#]`


The third sentence contains the original typo “The has placed”; it is retained so the supplied input is tested exactly.

````python
comparison_sentences = [
    "Did you search on Google?",
    "Let us google it.",
    "The has placed the food on the table",
    "She plans to table a motion for debate in Parliament"
]
comparison_results = []
for number, sentence in enumerate(comparison_sentences, start=1):
    sen = nlp(sentence)
    word_index = next(token.i for token in sen if token.text.lower() in {"google", "table"})
    word = sen[word_index]
    explanation = spacy.explain(word.tag_)
    comparison_results.append((sentence, word.text, word.pos_, word.tag_, explanation))
    print(f"Sentence {number}: {sentence}")
    print("Word:", word.text)
    print("Coarse-grained POS:", word.pos_)
    print("Fine-grained POS:", word.tag_)
    print("Explanation:", explanation)
    print()
````

### Task 2 comparison

In “Did you search on Google?”, **Google** names a specific service, so it functions as a proper noun (`PROPN`, `NNP`). In “Let us google it.”, **google** means to search online, so it functions as a base-form verb (`VERB`, `VB`). In the third sentence, **table** refers to a piece of furniture and is a singular common noun (`NOUN`, `NN`). In the fourth sentence, **table** describes presenting a motion for debate and is a base-form verb (`VERB`, `VB`). The same spelling can have a different grammatical role depending on context; statistical model predictions may occasionally differ from the intended analysis.
